# kalman filter

```python
from filterpy.kalman import KalmanFilter
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Initialize Kalman Filter
kf = KalmanFilter(dim_x=2, dim_z=1)
kf.x = np.array([[0.], [0.]])  # Initial state (position, velocity)
kf.F = np.array([[1., 1.], [0., 1.]])  # State transition matrix
kf.H = np.array([[1., 0.]])  # Measurement function
kf.P *= 1000  # Initial uncertainty
kf.R = 5  # Measurement noise
kf.Q = np.array([[1., 0.], [0., 1.]])  # Process noise
```

```python
fs = 100  # Sampling frequency (Hz)
t = np.linspace(0, 1, fs, endpoint=False)
signal = np.sin(2*np.pi*10*t) + np.sin(2*np.pi*50*t) + np.random.randn(len(t)) * 0.5  # Mixture of 10Hz & 50Hz + noise
kalman_data = []

for z in signal:
    kf.predict()
    kf.update(z)
    kalman_data.append(kf.x[0, 0])  # Store estimated position

# print(filtered_values)
```

```python
plt.plot(signal,label='data')
plt.plot(kalman_data,label='kalman smooth')
plt.legend()
plt.show()
```

# band pass filter

```python
from scipy.signal import butter, filtfilt
import numpy as np
```

```python
# Create a synthetic noisy signal
# fs = 100  # Sampling frequency (Hz)
# t = np.linspace(0, 1, fs, endpoint=False)
# signal = np.sin(2*np.pi*10*t) + np.sin(2*np.pi*50*t) + np.random.randn(len(t)) * 0.5  # Mixture of 10Hz & 50Hz + noise
```

```python
# Band-pass filter settings
lowcut, highcut = 8, 12  # Target frequency range (e.g., keeping only 10Hz)
b, a = butter(N=4, Wn=[lowcut / (fs / 2), highcut / (fs / 2)], btype='band')
```

```python
# Apply filter
bandpass_data = filtfilt(b, a, signal)

plt.plot(signal)
plt.plot(bandpass_data)
plt.show()
```

# Savitzky Golay filter

```python
from scipy.signal import savgol_filter
```

```python
savgol_data = savgol_filter(signal, window_length=5, polyorder=2)
```

```python
plt.plot(signal)
plt.plot(savgol_data)
plt.show()
```

# Median filter

```python
from scipy.signal import medfilt
```

```python
median_data = medfilt(signal, kernel_size=3)
```

```python
plt.plot(signal)
plt.plot(median_data)
plt.show()
```

# low pass filter

completely rounding the signal

```python
from scipy.signal import butter, filtfilt
```

```python
lowcut = 10.0
highcut = 20.0
fs = 100.0  # Sampling frequency

# Design filter
b, a = butter(4, lowcut / (fs / 2), btype='low')
```

```python
lowpass_data = filtfilt(b, a, signal)
```

```python
plt.plot(signal)
plt.plot(lowpass_data)
plt.show()
```

# high-pass filter

The other way. This one keeps the ruggedness of the data.

```python
highcut = 0.1
fs = 1.0  # Sampling frequency

# Design filter
b, a = butter(4, highcut / (fs / 2), btype='high')

# Apply the high-pass filter
highbass_data = filtfilt(b, a, signal)
```

```python
plt.plot(signal)
plt.plot(highbass_data)
plt.show()
```

# Wavelet Transform

```python
import pywt
```

```python
coeffs = pywt.wavedec(signal, 'haar', level=3)

# Threshold small coefficients (denoising)
threshold = 0.3
coeffs[1:] = [pywt.threshold(i, threshold, mode='soft') for i in coeffs[1:]]

# Reconstruct data
wavelet_data = pywt.waverec(coeffs, 'haar')
```

```python
plt.plot(signal)
plt.plot(wavelet_data)
plt.show()
```

# Wiener Filter

```python
from scipy.signal import wiener
```

```python
wiener_data = wiener(signal)
```

```python
plt.plot(signal)
plt.plot(wiener_data)
plt.show()
```

## signal to noise

signal to noise ratio considers that you already have the original (not noisy data) which we don't in real life applications. 

How can we use this idea?

```python
def snr(signal, noise):
    return 10 * np.log10(np.mean(signal**2) / np.mean(noise**2))
```

```python
kalman_snr = snr(signal, signal - kalman_data)
print(f"kalman SNR: {kalman_snr:.2f} dB")

bandpass_snr = snr(signal, signal - bandpass_data)
print(f"bandpass SNR: {bandpass_snr:.2f} dB")

savgol_snr = snr(signal, signal - savgol_data)
print(f"savgol SNR: {savgol_snr:.2f} dB")

median_snr = snr(signal, signal - median_data)
print(f"median SNR: {median_snr:.2f} dB")

lowpass_snr = snr(signal, signal - lowpass_data)
print(f"lowpass SNR: {lowpass_snr:.2f} dB")

highpass_snr = snr(signal, signal - highbass_data)
print(f"highpass SNR: {highpass_snr:.2f} dB")

wavelet_snr = snr(signal, signal - wavelet_data)
print(f"wavelet SNR: {wavelet_snr:.2f} dB")

wiener_snr = snr(signal, signal - wiener_data)
print(f"highpass SNR: {wiener_snr:.2f} dB")
```

```python
t = np.linspace(0, 2*np.pi, 100)  # Time vector
_signal = np.sin(2 * t)
```

```python
noise = np.random.normal(0, 0.3, size=_signal.shape)  # Noise with standard deviation 0.3
noisy_signal = _signal + noise
plt.plot(t,noisy_signal)
plt.plot(t,_signal)
plt.show()
```

```python
filtered_signal = savgol_filter(noisy_signal, window_length=9, polyorder=2)
```

```python
snr_before = snr(_signal, noise)
snr_after = snr(_signal, filtered_signal - _signal)
```

```python
# Step 4: Plot the results
plt.figure(figsize=(10, 5))

plt.plot(t, _signal, label="Original Signal", linestyle="dashed", linewidth=2)
plt.plot(t, noisy_signal, label=f"Noisy Signal (SNR: {snr_before:.2f} dB)", alpha=0.6)
plt.plot(t, filtered_signal, label=f"Filtered Signal (SNR: {snr_after:.2f} dB)", linewidth=2)

plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.legend()
plt.title("Effect of Filtering on Signal-to-Noise Ratio (SNR)")
plt.grid()
plt.show()
```
